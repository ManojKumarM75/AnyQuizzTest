```python


# =========================================================
# PROMPT
# =========================================================
prompt = """
Read the ENTIRE PDF carefully.

DO NOT skip any section, page, table, exercise,
example, activity, question, summary, or diagram text.

Generate MCQs with FULL COVERAGE of the PDF.

Requirements:
- Cover ALL topics from the PDF
- No important concept should be missed
- Generate as many MCQs as needed for full coverage
- 4 options per question
- Mention correct answer
- Add short explanation
- Avoid duplicates
- Output in SIMPLE MARKDOWN FORMAT ONLY

FORMAT:

# MCQ 1
Q. Question here

A. Option A
B. Option B
C. Option C
D. Option D

Answer: B
Explanation: ...

# MCQ 2
...

Do not output JSON.
Do not use code blocks.
"""

```

```python

# =========================================================
# GENERATE WITH RETRY
# =========================================================
for attempt in range(5):

    try:

        print(f"\nGemini Attempt {attempt+1}")

        response = client.models.generate_content(
            model="gemini-2.5-flash-lite",
            contents=[
                uploaded_file,
                prompt
            ]
        )

        output = response.text.strip()

        print("Generation Successful")

        break

    except Exception as e:

        print("Failed:", e)

        if attempt == 4:
            raise e

        time.sleep(10)

# =========================================================
# SAVE MD FILE
# =========================================================
md_file = "mcqs_output.md"

with open(md_file, "w", encoding="utf-8") as f:
    f.write(output)

# =========================================================
# COUNT QUESTIONS
# =========================================================
mcq_count = len(re.findall(r"# MCQ", output))

print("\n==============================")
print("TOTAL MCQs GENERATED:", mcq_count)
print("==============================")

# =========================================================
# PREVIEW
# =========================================================
print("\nFIRST 5000 CHARACTERS:\n")
print(output[:5000])

# =========================================================
# DOWNLOAD FILE
# =========================================================
files.download(md_file)
```
